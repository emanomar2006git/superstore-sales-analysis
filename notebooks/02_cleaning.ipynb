## Setup

Importing required libraries.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

## Load Raw Data

Loading raw CSV untouched from Kaggle. Check shape, dtypes, and missing values before cleaning.

```python
df = pd.read_csv("../data/superstore.csv", encoding="latin-1")  # from notebooks/ folder
print(f"Loaded {df.shape[0]} rows, {df.shape[1]} columns")
df.head()
```

```python
print(df.dtypes)
print()
print(df.isnull().sum())
print(f"Duplicates: {df.duplicated().sum()}")
```

## Cleaning Step 1 — Parse Dates

`Order Date` and `Ship Date` are stored as strings (e.g., '11/8/2016'). Convert to datetime and verify range 2014-01-03 to 2017-12-30.

```python
df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")
df["Ship Date"] = pd.to_datetime(df["Ship Date"], errors="coerce")
print(df[["Order Date", "Ship Date"]].dtypes)
print(pd.to_datetime(df["Order Date"]).min(), "to", pd.to_datetime(df["Order Date"]).max())
```

## Cleaning Step 2 — Trim Whitespace in Categorical Columns

Category / Region / Segment names may have leading/trailing spaces. Strip them and verify unique values.

```python
for col in ["Category", "Sub-Category", "Region", "Segment", "Ship Mode"]:
    df[col] = df[col].astype(str).str.strip()
print(df["Category"].unique())
print(df["Region"].unique())
```

## Cleaning Step 3 — Check for Duplicates

Look for duplicate rows and decide whether to drop them. Document decision.

```python
print(f"Duplicate rows: {df.duplicated().sum()}")
df = df.drop_duplicates()
print(f"After dropping duplicates: {df.shape[0]} rows")
```

## Cleaning Step 4 — Numeric Columns Stored as Strings

Verify Sales, Profit, Discount, Quantity are numeric. If any are strings (e.g., with commas), coerce to numeric.

```python
print(df[["Sales", "Quantity", "Discount", "Profit"]].dtypes)
for col in ["Sales", "Quantity", "Discount", "Profit"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")
print("After coercion:")
print(df[["Sales", "Quantity", "Discount", "Profit"]].dtypes)
```

## Cleaning Step 5 — Handle Unwanted Columns

`Row ID` is often an artifact. **Mentor decision (Q6): KEEP Row ID for traceability** — do not drop it, just document that you kept it and why.

```python
# Decision: KEEP Row ID (do not drop) — it helps trace back to raw data
# No drop needed. Just verify columns:
print(df.columns.tolist())  # should still include 'Row ID'
```

## Cleaning Step 6 — Final Checks

Re-check dtypes, missing values, shape, and that dates parsed correctly.

```python
print(df.dtypes)
print()
print(df.isnull().sum())
print()
print(df.describe())
print()
print(f"Final shape: {df.shape}")
print(f"Date range: {df['Order Date'].min()} to {df['Order Date'].max()}")
```

## Save Cleaned Data

Save to `data/superstore_clean.csv` so the main notebook can load it reproducibly.

```python
df.to_csv("../data/superstore_clean.csv", index=False)
print("Saved to ../data/superstore_clean.csv")
```
