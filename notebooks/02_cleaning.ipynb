## Cleaning Step 1 — Parse Dates

`Order Date` and `Ship Date` are stored as strings (e.g., '11/8/2016'). Convert to datetime and verify range 2014-01-03 to 2017-12-30.

```python
# Convert Order Date and Ship Date to datetime
df["Order Date"] = pd.to_datetime(df["Order Date"], errors="coerce")
df["Ship Date"] = pd.to_datetime(df["Ship Date"], errors="coerce")
print(df[["Order Date", "Ship Date"]].dtypes)
print(df["Order Date"].min(), "to", df["Order Date"].max())
```

## Cleaning Step 2 — Trim Whitespace in Categorical Columns

Category / Region / Segment names may have leading/trailing spaces. Strip them and verify unique values.

```python
#  Strip whitespace for key categorical columns
for col in ["Category", "Sub-Category", "Region", "Segment", "Ship Mode"]:
    df[col] = df[col].astype(str).str.strip()
print(df["Category"].unique())
print(df["Region"].unique())
```
